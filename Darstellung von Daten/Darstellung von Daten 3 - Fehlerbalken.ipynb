# Darstellung von Daten 3 - Fehlerbalken

## Einleitung

Wie wir in einer vorherigen Lektion gesehen haben, sind Messungen immer mit Messunsicherheiten versehen. Wir haben bereits gelernt, wie wir dies bei einzelnen Messungen angeben können, z.B.

$t = (9.85 \pm 0.18)$ s

Wenn wir nun jedoch mehrere Messwerte haben und diese in einem Diagramm darstellen wollen, so benötigen wir eine grafische Repräsentation der Messunsicherheiten. Hierfür gibt es die sogenannten Fehlerbalken.



## Wiederholung des Beispiels

Wir nutzen noch einmal die Beispieldaten aus "Darstellung von Daten 2 - Punktdiagramme". Führe folgenden Code aus, um die Daten zu laden und nochmals das Diagramm zu erzeugen.

```python
# Modul zum Handling von Datendateien
import pandas as pd

# Modul zur Darstellung von Daten
import matplotlib.pyplot as plt

# Daten aus Datei einlesen
data = pd.read_csv("sprint60m_mitgroessen.txt")

# Daten zur Kontrolle ausgeben
print(data)

# Datenreihen in Listen umwandeln für spätere Verwendung
times = data["Zeiten in s"].to_list()
height = data["Grösse in m"].to_list()

# Diagrammvariablen erstellen
fig, ax = plt.subplots()

# Punktdiagramm erstellen
ax.scatter(height, times, color="violet", s=20, marker="o") # Erstellung des Punktdiagramms mit den Daten, der Farbe, Grösse der Datenpunkte und deren Symbol (z.B. "o", "x", "d", "*")

# Diagrammgrenzen
ax.set_xlim(1.54, 1.69)
ax.set_ylim(9, 11)

# Achsen beschriften
ax.set_xlabel('Grösse in m')
ax.set_ylabel('Zeit in s')

#Diagramm anzeigen
plt.show()
```

## Erzeugen der Fehlerbalken

Für die Fehlerbalken gibt es in Matplotlib auch eine eingebaute Funktion. Wir müssen nur zusätzlich für jeden Datenpunkt angeben, wie gross die Unsicherheit ist. Testen Sie folgenden Code und veränderen Sie dann die Grösse der Balken, um den Code zu verstehen.

```python
# Diagrammvariablen erstellen
fig, ax = plt.subplots()

# Punktdiagramm erstellen
ax.scatter(height, times, color="violet", s=20, marker="o") # Erstellung des Punktdiagramms mit den Daten, der Farbe, Grösse der Datenpunkte und deren Symbol (z.B. "o", "x", "d", "*")

# Diagrammgrenzen
ax.set_xlim(1.54, 1.69)
ax.set_ylim(9, 11)

# Achsen beschriften
ax.set_xlabel('Grösse in m')
ax.set_ylabel('Zeit in s')

# Messunsicherheiten definieren
t_uncertain = []
for i in range(len(times)):
    t_uncertain.append(0.2)

# Fehlerbalken hinzufügen
ax.errorbar(height, times, yerr=t_uncertain, xerr=None, fmt='none', color="violet", capsize=3)

# Diagramm anzeigen
plt.show()
```
